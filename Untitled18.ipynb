```python
validation_set=tf.keras.utils.image_dataset_from_directory(
    '/content/New Plant Diseases Dataset(Augmented)/New Plant Diseases Dataset(Augmented)/valid',
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
)
```

```python
test_set=tf.keras.utils.image_dataset_from_directory(
    '/content/New Plant Diseases Dataset(Augmented)/New Plant Diseases Dataset(Augmented)/valid',
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=False,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
)
```

```python
import matplotlib.pyplot as plt

# Function to scale the images
def scale(image, label):
    image = tf.cast(image, tf.float32) / 255.0  # Scale image to [0, 1] range
    return image, label

# Apply scaling
training_set = training_set.map(scale)

# Take one batch of data
for images, labels in training_set.take(1):
    # images is a batch of images, labels is a batch of labels
    print(f'Images shape: {images.shape}')
    print(f'Labels shape: {labels.shape}')

    # Extract a single image and label
    image = images[0].numpy()  # Convert the first image to a numpy array
    label = labels[0].numpy()  # Convert the first label to a numpy array

    # Print the image and label
    print(f'Single image shape: {image.shape}')
    print(f'Single label: {label}')
    print(f'Single image pixel values range: [{image.min()}, {image.max()}]')

    # Display the image using matplotlib
    plt.imshow(image)
    plt.title(f'Label: {label}')
    plt.axis('off')
    plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization,GlobalAveragePooling2D,LeakyReLU,SeparableConv2D
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, LearningRateScheduler
```

```python
# Callbacks
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=1)
```

```python
tf.random.set_seed(42)

model = Sequential()

model.add(Conv2D(filters=64,kernel_size=3,padding='same',input_shape=(128,128,3),activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=2,strides=2))

model.add(SeparableConv2D(filters=128,kernel_size=3,padding='same',activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=2,strides=2))

model.add(Conv2D(filters=256,kernel_size=3,padding='same',activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=2,strides=2))

model.add(SeparableConv2D(filters=512,kernel_size=3,padding='same',activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=2,strides=2))

model.add(Conv2D(filters=1024,kernel_size=3,padding='same',activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=2,strides=2))

model.add(GlobalAveragePooling2D())

model.add(Dense(units=1024, kernel_regularizer=l2(0.01)))
model.add(BatchNormalization())
model.add(LeakyReLU(alpha=0.01))
model.add(Dropout(0.5))

model.add(Dense(units=512, kernel_regularizer=l2(0.01)))
model.add(BatchNormalization())
model.add(LeakyReLU(alpha=0.01))
model.add(Dropout(0.5))

model.add(Dense(units=38, activation='softmax'))
```

```python
#compile
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
training_history=model.fit(training_set,epochs=10,validation_data=validation_set,callbacks=[early_stopping, reduce_lr])
```

```python
import matplotlib.pyplot as plt
epochs = [i for i in range(1,11)]
plt.plot(epochs,training_history.history['accuracy'],color='red',label='Training Accuracy')
plt.plot(epochs,training_history.history['val_accuracy'],color='blue',label='Validation Accuracy')
plt.xlabel('No. of epochs')
plt.ylabel('Accuracy Result')
plt.title('Visualization of Accuracy Result')
plt.legend()
plt.show()
```

```python
import numpy as np
for layer in model.layers:
        if isinstance(layer, Dense):
            weights, biases = layer.get_weights()
            threshold = np.percentile(np.abs(weights), 20)
            pruned_weights = np.where(np.abs(weights) < threshold, 0, weights)
            layer.set_weights([pruned_weights, biases])
```

```python
class_name = validation_set.class_names
class_name
```

```python
y_pred = model.predict(test_set)
```

```python
predicted_categories = tf.argmax(y_pred,axis=1)
```

```python
true_categories = tf.concat([y for x,y in test_set],axis=0)
true_categories
```

```python
y_true = tf.argmax(true_categories,axis=1)
y_true
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_true,predicted_categories,target_names=class_name))
```

```python
cm = confusion_matrix(y_true,predicted_categories)
cm
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(40,40))
sns.heatmap(cm,annot=True,annot_kws={'size':10})
plt.xlabel('Predicted Class',fontsize=20)
plt.ylabel('Actual Class',fontsize=20)
plt.title('Plant Disease Prediction Confusion Matrix',fontsize=30)
plt.show()
```
